```python
# %% Intent [markdown]
# Take models, prepare them for use with mergem, and then use mergem to extend all annotations.

# %%
# Library imports

# Hack to allow importing local libraries inside notebook.
import sys
import os
sys.path.append(os.getcwd() + '/..')

import re
from collections import defaultdict
from pathlib import Path
from cobra.io import read_sbml_model
from pandas import Series, DataFrame
import requests
from cobra import Model, Reaction, Metabolite
from cobra.io import write_sbml_model
import mergem
import model_utils as pm_mu

# %%
# Extend a chosen model

# Input
input_path = "../Model/250327_iABA974.sbml"
output_file = "../Model/250409_iABA974_mergem.sbml"

summaries = pm_mu.ModelSummary()
annotstats = pm_mu.ModelAnnotationStats()
print(f"Importing model from: {input_path}")
orig_model = read_sbml_model(input_path)
summaries.add(orig_model, "original model")
annotstats.add(orig_model, "original model")
orig_model = pm_mu.prep_sof_model_for_merge(orig_model)
mr = mergem.merge( input_models=[orig_model], extend_annot=True )
ext_model = mr["merged_model"]
summaries.add(ext_model, "extended model")
annotstats.add(ext_model, "extended model")
write_sbml_model(ext_model, output_file)
print(f"Extended model written to: {output_file}")
print(summaries.output())
print(annotstats.output())

```
