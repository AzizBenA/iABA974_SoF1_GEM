Create reaction flux csv

```python
reactions_data = pd.DataFrame(columns=['ID','time 0sec'])
solution = model.optimize()
# Iterate over all reactions in the model and populate the DataFrame
for reaction in model.reactions:
    # Create a DataFrame for the current reaction
    reaction_data = pd.DataFrame([{
        "ID": reaction.id,
        "time 0sec": solution.fluxes[reaction.id],}])
    
    # Concatenate the current reaction's DataFrame to the main DataFrame
    reactions_data = pd.concat([reactions_data, reaction_data], ignore_index=True)

# Print the resulting DataFrame to verify the results
reactions_data.to_csv('../Data/Escher_data/reactions_data_BLG.csv',index=False)
```
