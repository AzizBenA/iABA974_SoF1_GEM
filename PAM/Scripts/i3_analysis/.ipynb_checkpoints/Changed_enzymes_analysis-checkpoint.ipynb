```python
import os
from matplotlib.colors import to_hex
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import matplotlib.colors as mcolors
import seaborn as sns
import pandas as pd
import numpy as np
import json
import re
import cobra
from scipy.stats import entropy
import scipy.cluster.hierarchy as sch
from scipy.cluster.hierarchy import fcluster
from typing import Literal
from itertools import product


from matplotlib.colors import LogNorm, SymLogNorm
from sklearn.preprocessing import MinMaxScaler


if os.path.split(os.getcwd())[1] == 'i3_analysis':
    os.chdir('../..')
    
from Modules.utils.pam_generation import (get_rxn2kcat_protein2gene_dict, 
                                          _extract_reaction_id_from_catalytic_reaction_id,
                                          _get_rxn2kcat_as_series
                                         )
from Modules.utils.pamparametrizer_analysis import (calculate_kcat_differences, 
                                                    plot_histogram_logspace,
                                                    select_clustered_rows_by_variation,
                                                    get_clusters_from_clustermap
                                                   )
N_ALT_MODELS = 8

ECOLI_MODEL_PATH = os.path.join('Models', 'iML1515.xml')
DIAGNOSTIC_FILES = [os.path.join('Results','2_parametrization', 'diagnostics',
                                 f'pam_parametrizer_diagnostics_{file_nmbr}.xlsx')for file_nmbr in range(1,N_ALT_MODELS+1)]
PAM_DATA_FILE_PATH = os.path.join('Results','2_parametrization', 'proteinAllocationModel_iML1515_EnzymaticData_multi.xlsx')
PAM_KCAT_FILES = [os.path.join('Results','3_analysis', 'parameter_files',
                                 f'proteinAllocationModel_EnzymaticData_iML1515_{file_nmbr}.xlsx')for file_nmbr in range(1,N_ALT_MODELS+1)]

max_iteration = 8
```

## 1. Get the data
### 1.1 Get the original kcat values

```python
original_kcat_df = pd.read_excel(PAM_DATA_FILE_PATH, sheet_name='ActiveEnzymes')
original_kcat_df
```

```python
# setup the gem ecoli iML1515 model
rxn2kcat, _ = get_rxn2kcat_protein2gene_dict(PAM_DATA_FILE_PATH, ECOLI_MODEL_PATH)
```

### 1.2 Get the diagnostics files

```python
best_individual_df = None

for i, file in enumerate(DIAGNOSTIC_FILES):
    pam_param_results = pd.read_excel(file, sheet_name = 'Best_Individuals')
    pam_param_error = pd.read_excel(file, sheet_name='Final_Errors')
    pam_param_results = pam_param_results.merge(pam_param_error, on='run_id', how='left')
    pam_param_results['alternative'] = i+1
    
    if best_individual_df is None: 
        best_individual_df = pam_param_results
    else:
        best_individual_df = pd.concat([best_individual_df,pam_param_results])
best_individual_df
```

### 1.3. Get the altered kcat values

```python
best_individual_df_catalytic_rxn = best_individual_df.copy()
best_individual_df['rxn_id'] = best_individual_df['rxn_id'].apply(lambda row: 
                                                                        _extract_reaction_id_from_catalytic_reaction_id(row))

best_individual_df
```

## 2. Calculate the relative change in kcat
### 2.1 Match the original kcats

```python
# Ensure the dataframe is sorted by enzyme and run
best_indiv_all_rounds_diff_kcat = best_individual_df.groupby(
    'alternative', group_keys=True).apply(calculate_kcat_differences, rxn2kcat).reset_index(drop=True)
best_indiv_all_rounds_diff_kcat.sort_values(by='kcat_change')
```

### 3.5 analysis of enzymes with the most changes

```python
def determine_number_nonzero_changes(best_indiv_df: pd.DataFrame, groups: list[str])-> pd.DataFrame:
    changed_kcat_counts = (
    best_indiv_df.groupby(groups).agg(
        nonzero_count=('kcat_change', lambda x: (x != 0).sum()),  # Count nonzero entries
        positive_count=('kcat_change', lambda x: (x > 0).sum()),  # Count positive entries
        negative_count=('kcat_change', lambda x: (x < 0).sum()),  # Count negative entries
    ).reset_index()
    )
#     changed_kcat_count = pd.merge(changed_kcat_counts, best_indiv_df)
    return changed_kcat_counts.sort_values(by = 'nonzero_count',ascending=False)
```

```python
# Group by the unique combination of enzyme_id, direction, and rxn_id
changed_kcat_counts = determine_number_nonzero_changes(best_indiv_for_heatmap[best_indiv_for_heatmap.run_id<4],
                                                       ['enzyme_id', 'direction', 'rxn_id'])
changed_kcat_counts
```

```python
# Group by the unique combination direction and rxn_id
changed_rxn_counts = determine_number_nonzero_changes(best_indiv_for_heatmap[best_indiv_for_heatmap.run_id<4],
                                                      ['direction', 'rxn_id'])
changed_rxn_counts
```

```python
# Group by the unique combination enzyme
changed_enz_counts = determine_number_nonzero_changes(best_indiv_for_heatmap[best_indiv_for_heatmap.run_id<4], ['enzyme_id'])
changed_enz_counts
```

### 4. Changes per pathway

### 4.1 Merging the information

```python
# get pathway information from genelist file
gene_list_info = pd.read_excel(os.path.join('Data', 'GeneList_ecoli.xlsx'), sheet_name = 'GeneList')
rxn2gene_info = pd.read_excel(os.path.join('Data', 'GeneList_ecoli.xlsx'), sheet_name = 'gene2rxn')
```

```python
# map cog information to reactions using the gene id
gene2rxn2cog = pd.merge(gene_list_info[['bnumber', 'COG description']], rxn2gene_info,
                       left_on = 'bnumber', right_on = 'Gene', how = 'inner')
gene2rxn2cog.Reactions = gene2rxn2cog.Reactions.str.split(', ')
gene2rxn2cog['COG description'] = gene2rxn2cog['COG description'].str.split(';')
gene2rxn2cog = gene2rxn2cog.explode('Reactions')
gene2rxn2cog = gene2rxn2cog.explode('COG description').reset_index(drop=True)

gene2rxn2cog
```

```python
# map cog information to changed enzymes using the reactions
best_kcats_with_cog = pd.merge(best_indiv_all_rounds_diff_kcat, gene2rxn2cog[['Reactions', 'COG description']],
                               left_on='rxn_id', right_on = 'Reactions', how='left')
best_kcats_with_cog = best_kcats_with_cog.drop_duplicates(
    ['rxn_id','enzyme_id', 'r_squared', 'COG description']).reset_index(drop=True)

best_kcats_with_cog
```

```python
# Calculate cumulative change of each COG category per iteration
best_kcats_change_per_cog = best_kcats_with_cog.groupby(['alternative', 'COG description'])['relative_change'].sum().reset_index()

# Calculate the absolute sum of relative changes per iteration
total_abs_change_per_iteration = best_kcats_with_cog.groupby('alternative')['relative_change'].sum().abs().reset_index()
total_abs_change_per_iteration.rename(columns={'relative_change': 'total_abs_change'}, inplace=True)

# Calculate the absolute sum of changes per iteration
total_change_per_iteration = best_kcats_with_cog.groupby(['alternative', 'COG description'])['kcat_change'].sum().reset_index()
total_change_per_iteration.rename(columns={'kcat_change': 'total_change'}, inplace=True)

counts_per_cog_iteration = best_kcats_with_cog.groupby(['COG description', 'alternative']).count()[['Reactions']]

# avg_change_per_iteration['avg_change'] = avg_change_per_iteration['total_change']/avg_change_per_iteration['Reactions']


# Merge the cumulative changes with the total absolute changes
best_kcats_change_per_cog = pd.merge(best_kcats_change_per_cog, total_abs_change_per_iteration, on='alternative')
best_kcats_change_per_cog = pd.merge(best_kcats_change_per_cog, total_change_per_iteration, on=['alternative', 'COG description'])
best_kcats_change_per_cog = pd.merge(best_kcats_change_per_cog, counts_per_cog_iteration, on=['alternative', 'COG description'])

# Normalize the cumulative changes by the total absolute changes within each iteration
best_kcats_change_per_cog['normalized_relative_change'] =best_kcats_change_per_cog['relative_change'] / best_kcats_change_per_cog['total_abs_change']
best_kcats_change_per_cog['avg_change'] = best_kcats_change_per_cog['total_change']/best_kcats_change_per_cog['Reactions']


best_kcats_change_per_cog.drop('Reactions', axis=1, inplace=True)
```

```python
# plot the relative changes
cog_descriptions = best_kcats_change_per_cog['COG description'].drop_duplicates().to_list()
cog2rxns = {}
for i, row in best_kcats_with_cog[['Reactions', 'COG description']].iterrows():
    cog = row['COG description']
    if cog in cog2rxns.keys():
        cog2rxns[cog] += [row.Reactions]
    else:
        cog2rxns[cog] = [row.Reactions]
```

```python
def plot_histogram_logspace(ax: plt.Axes, data:pd.DataFrame,
                            color,
                            label: str, x_label:str, n_bins:int=50,
                            relative:bool=False, ymax: float = None) -> None:
    if not relative:
        kwargs = {}
        ax.set_yscale('log')
        if not ymax: ymax = 1e3
    else:
        kwargs = {'weights': np.ones_like(data) / len(data)}
        if not ymax: ymax = 0.35

    hist, bins = np.histogram(data, bins=n_bins, **kwargs)
    logbins = np.logspace(np.log10(bins[0]), np.log10(bins[-1]), len(bins))

    bin_heights, bin_borders, _ = ax.hist(data, bins=logbins, histtype='step',
                                          stacked=True, fill=False, label=label, color=color, **kwargs)

    ax.vlines(13.7, 0, ymax, linestyle='dotted')
    ax.set_ylim([0, ymax])

    ax.set_xlabel(x_label)
    ax.set_xscale('log')

```

```python
def plot_PCA_graph(input_df, columns_to_analyse: list[str], hue:str, values:str):
    # Pivot the data to have alternatives as rows and enzymes as columns
    # This creates a wide matrix required for PCA
    data_matrix = input_df.pivot_table(
        index=columns_to_analyse,
        columns=hue,
        values=values,
        aggfunc='mean',
        fill_value=0
    )

    data_matrix.reset_index(names=columns_to_analyse, inplace=True)
    mapping_to_analyse = dict()
    for col in columns_to_analyse:
        mapping_to_analyse[col] = data_matrix[col]

    data_for_pca = data_matrix.drop(columns=columns_to_analyse)

    # Apply PCA
    pca = PCA(n_components=2)
    reduced_data = pca.fit_transform(data_for_pca)
    explained_variance = pca.explained_variance_ratio_
    print(f"PCA explained variance: {explained_variance}")

    # Create a DataFrame for plotting
    reduced_df = pd.DataFrame(reduced_data, columns=['PC1', 'PC2'])

    for col, col_series in mapping_to_analyse.items():
        reduced_df[col] = col_series

    # Merge back descriptions for coloring
    mapping = input_df[columns_to_analyse+[hue]].drop_duplicates()
    reduced_df = pd.merge(reduced_df, mapping, on=columns_to_analyse, how='left')

    # Plot the reduced data
    plt.figure(figsize=(12, 8))
    sns.scatterplot(
        x='PC1', y='PC2', hue='COG description', data=reduced_df, palette='tab20', s=100, alpha=0.8
    )

    # Add titles and labels
    plt.title('PCA of Enzyme Changes per Alternative', fontsize=16)
    plt.xlabel(f'Principal Component 1 ({round(explained_variance[0], 3) * 100}%)', fontsize=12)
    plt.ylabel(f'Principal Component 2 ({round(explained_variance[1], 3) * 100}%)', fontsize=12)

    # Position the legend outside the plot
    plt.legend(
        bbox_to_anchor=(1.05, 0.5),  # Position the legend to the right-hand side, centered vertically
        loc='center left',  # Align it to the left of the bounding box
        title=hue
    )
    plt.tight_layout()
#     plt.show()
    return fig, ax
```
